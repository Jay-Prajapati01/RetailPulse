# RetailPulse | Day 3 Customer Segmentation

This notebook builds a customer segmentation pipeline on top of the processed Online Retail II data. It uses the engineered RFM table, scales the customer features, fits KMeans and DBSCAN, derives customer personas, and saves the trained models for reuse in RetailPulse.

## Workflow

1. Load the processed RFM dataset.
2. Scale Recency, Frequency, and Monetary features.
3. Evaluate KMeans with the elbow method and silhouette score.
4. Fit DBSCAN for density-based segmentation.
5. Create PCA and t-SNE visualizations.
6. Map clusters to customer personas and save models.

```python
from pathlib import Path

import pandas as pd

from retailpulse_customer_segmentation import run_customer_segmentation_pipeline
from retailpulse_feature_pipeline import PROCESSED_DIR, MODELS_DIR, FIGURES_DIR

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)

print('Processed data folder:', PROCESSED_DIR)
print('Model folder:', MODELS_DIR)
print('Figure folder:', FIGURES_DIR)
```

```python
clustered_customers = run_customer_segmentation_pipeline()
clustered_customers.head()
```

```python
cluster_summary = clustered_customers.groupby('kmeans_cluster')[['Recency', 'Frequency', 'Monetary']].mean().round(2)
persona_summary = clustered_customers['persona'].value_counts().to_frame('customer_count')
display(cluster_summary)
display(persona_summary)
```

**Business insight:** RetailPulse should treat Champions and Loyal Customers as the primary retention and upsell audience, while At Risk and Lost customers should trigger win-back, service recovery, and churn-prevention campaigns. Cluster-level behavior should be reviewed after each promotion cycle because retail customer mixes move quickly.
